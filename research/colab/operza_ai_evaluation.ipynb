# Operza Enterprise AI & Copilot Quality Evaluation
Evaluates tool calling accuracy, parameter validation, and hallucination rates across synthetic agency scenarios.

```python
import json
import pandas as pd

scenarios = [
    {
        'prompt': 'Create a high-value opportunity for Wayne Enterprises worth $50,000',
        'expected_tool': 'crm_create_opportunity',
        'expected_params': {'title': 'Enterprise Deal', 'value': 50000},
        'actual_tool': 'crm_create_opportunity',
        'actual_params': {'title': 'Wayne Enterprises Deal', 'value': 50000},
        'hallucination': False
    },
    {
        'prompt': 'Send an email to John Doe saying our onboarding call is confirmed',
        'expected_tool': 'conversations_send_message',
        'expected_params': {'channel': 'email'},
        'actual_tool': 'conversations_send_message',
        'actual_params': {'channel': 'email'},
        'hallucination': False
    },
    {
        'prompt': 'What is our total revenue this month?',
        'expected_tool': 'reporting_revenue_summary',
        'expected_params': {},
        'actual_tool': 'reporting_revenue_summary',
        'actual_params': {},
        'hallucination': False
    }
]

df_eval = pd.DataFrame(scenarios)
tool_accuracy = (df_eval['expected_tool'] == df_eval['actual_tool']).mean() * 100
hallucination_rate = df_eval['hallucination'].mean() * 100

print('=== OPERZA COPILOT EVALUATION SCORECARD ===')
print(f'Total Scenarios Tested: {len(df_eval)}')
print(f'Tool Routing Accuracy: {tool_accuracy:.1f}%')
print(f'Hallucination Rate: {hallucination_rate:.1f}%')
print(df_eval[['prompt', 'expected_tool', 'actual_tool']])
```
