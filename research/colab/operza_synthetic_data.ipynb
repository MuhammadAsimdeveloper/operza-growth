# Operza Synthetic Test Data Generator
Generates realistic, completely non-identifying synthetic CRM contacts, opportunities, and workflows for testing.

```python
import json
import random
import uuid
from datetime import datetime

first_names = ['Alex', 'Jordan', 'Taylor', 'Morgan', 'Casey', 'Sam', 'Riley', 'Avery']
last_names = ['Vance', 'Sterling', 'Mercer', 'Hayden', 'Cross', 'Winters', 'Sinclair', 'Colby']
companies = ['Apex Global', 'Zenith Logistics', 'Summit Digital', 'Beacon Media', 'Nova Systems']

contacts = []
for i in range(25):
    fn = random.choice(first_names)
    ln = random.choice(last_names)
    comp = random.choice(companies)
    contacts.append({
        'id': str(uuid.uuid4()),
        'firstName': fn,
        'lastName': ln,
        'email': f'{fn.lower()}.{ln.lower()}_{i}@synthetic-operza.test',
        'company': comp,
        'phone': f'+1555{random.randint(1000000, 9999999)}',
        'status': random.choice(['lead', 'customer', 'prospect']),
        'tags': ['synthetic-dataset', comp.lower().replace(' ', '-')]
    })

print(f'Generated {len(contacts)} synthetic contacts.')
print(json.dumps(contacts[:3], indent=2))

with open('synthetic_contacts_fixture.json', 'w') as f:
    json.dump(contacts, f, indent=2)
print('Exported to synthetic_contacts_fixture.json successfully.')
```
